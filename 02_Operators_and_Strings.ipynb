## 1. Arithmetic operator

| ตัวดำเนินการ | ความหมาย | ตัวอย่าง | ผลลัพธ์ |
|---|---|---|---|
| `+ - * /` | บวก ลบ คูณ หาร | `7 / 2` | `3.5` |
| `//` | หารปัดลง (floor division) | `7 // 2` | `3` |
| `%` | เศษจากการหาร (modulo) | `7 % 2` | `1` |
| `**` | ยกกำลัง | `2 ** 10` | `1024` |

`%` มีประโยชน์มากในงานภูมิศาสตร์ เช่น หาว่าปีไหนเป็นปีอธิกสุรทิน หรือแปลงองศาให้อยู่ในช่วง 0–360

````python
rain_total = 1435.8
days = 365

print(rain_total / days)      # ค่าเฉลี่ยต่อวัน
print(rain_total // days)     # ปัดลงเป็นจำนวนเต็ม
print(rain_total % days)
print(2 ** 10)

# ลำดับการคำนวณเหมือนคณิตศาสตร์: () -> ** -> * / -> + -
print(2 + 3 * 4)
print((2 + 3) * 4)
````

````python
# ตัวอย่างการใช้ % : แปลงทิศลมที่เกิน 360 องศาให้กลับเข้าช่วง 0-360
wind_dir = 405
print(wind_dir % 360)

# เช็คปีอธิกสุรทินอย่างง่าย
year = 2024
print(year % 4 == 0)
````

## 2. Comparison และ Logical operator

ผลลัพธ์ของการเปรียบเทียบคือ `bool` (`True` / `False`) เสมอ — นี่คือหัวใจของบทถัดไป (`if`)

| เปรียบเทียบ | `==` เท่ากับ, `!=` ไม่เท่า, `>` `<` `>=` `<=` |
|---|---|
| ตรรกะ | `and` (และ), `or` (หรือ), `not` (ไม่) |

⚠️ `=` คือกำหนดค่า ส่วน `==` คือเปรียบเทียบ — นักศึกษาปี 2 พลาดตรงนี้กันทุกปี

````python
rain_mm = 45.5

print(rain_mm > 35)
print(rain_mm == 45.5)
print(rain_mm != 0)

is_heavy = rain_mm > 35 and rain_mm <= 90
print("ฝนหนัก:", is_heavy)

print(35 < rain_mm <= 90)   # Python เขียนต่อกันแบบนี้ได้เลย
````

## 3. ข้อความ (string) — index และ slicing

ข้อความคือลำดับของตัวอักษร นับตำแหน่งเริ่มจาก **0**
`text[start:stop]` จะได้ตั้งแต่ `start` ถึง **ก่อน** `stop`

````python
code = "CHB01_20250914_rain.csv"

print(len(code))
print(code[0])       # ตัวแรก
print(code[-1])      # ตัวสุดท้าย
print(code[0:5])     # CHB01
print(code[:5])      # เหมือนกัน ละ 0 ได้
print(code[6:14])    # 20250914
print(code[-3:])     # csv
````

## 4. เมธอดของ string ที่ใช้บ่อย

````python
raw = "   ChonBuri Station   "

print(raw.strip())            # ตัดช่องว่างหัวท้าย (ใช้บ่อยมากตอนอ่านไฟล์)
print(raw.strip().upper())
print(raw.strip().lower())
print(raw.strip().replace(" ", "_"))
print("Buri" in raw)          # ค้นหาข้อความย่อย
print(raw.strip().startswith("Chon"))
````

### `split()` และ `join()` — แกะและประกอบข้อความ

`split()` ตัดข้อความเป็น **list** ตามตัวคั่น ส่วน `join()` ทำกลับกัน
นี่คือพื้นฐานของการอ่านไฟล์ CSV ด้วยมือ (ซึ่งเราจะทำในวันที่ 5)

````python
filename = "CHB01_20250914_rain.csv"
parts = filename.replace(".csv", "").split("_")
print(parts)

station, datestr, variable = parts
print(f"สถานี = {station}")
print(f"วันที่ = {datestr[6:8]}/{datestr[4:6]}/{datestr[0:4]}")
print(f"ตัวแปร = {variable}")

# ประกอบกลับ
print("_".join(["SRC01", "20251002", "temp"]) + ".csv")

# แถวข้อมูลแบบ CSV
row = "2025-09-14,CHB01,118.6"
fields = row.split(",")
print(fields)
print(float(fields[2]) / 25.4, "inch")
````
