## 2. เมธอดของ list ที่ใช้บ่อย

| เมธอด | ทำอะไร |
|---|---|
| `append(x)` | เพิ่มต่อท้าย |
| `insert(i, x)` | แทรกที่ตำแหน่ง i |
| `remove(x)` | ลบค่าตัวแรกที่เจอ |
| `pop(i)` | ดึงค่าออกมาพร้อมลบ |
| `sort()` | เรียงลำดับ (แก้ list เดิม) |
| `reverse()` | กลับลำดับ |
| `index(x)` | หาตำแหน่งของค่า |
| `count(x)` | นับจำนวนครั้งที่พบ |

````python
s = ["CHB01", "SRC01", "NYI01"]

s.append("STH01")
print(s)

s.insert(0, "PTY01")
print(s)

s.remove("SRC01")
print(s)

last = s.pop()
print("ดึงออกมา:", last, "| เหลือ:", s)

s.sort()
print("เรียงแล้ว:", s)

print("มี NYI01 ไหม:", "NYI01" in s)
print("อยู่ตำแหน่ง:", s.index("NYI01"))
````

````python
# ✍️ พิมพ์โค้ดด้านบนด้วยตัวเองที่นี่
````

````python
values = [45.5, 0.0, 118.6, 12.4, 88.9]

print(sorted(values))                      # sorted() คืนค่าใหม่ ไม่แก้ของเดิม
print(sorted(values, reverse=True))
print(values)                              # ของเดิมยังอยู่ครบ
print(sum(values), max(values), min(values))
print(f"เฉลี่ย {sum(values)/len(values):.2f}")
````

## 3. List comprehension — เขียนลูปให้สั้นลง

รูปแบบ: `[ ค่าที่ต้องการ for ตัวแปร in ชุดข้อมูล if เงื่อนไข ]`

อ่านจากซ้ายไปขวาว่า "เอา *ค่าที่ต้องการ* สำหรับทุก *ตัวแปร* ใน *ชุดข้อมูล* ที่ *เงื่อนไข* เป็นจริง"

````python
rain = [0.0, 12.4, 0.0, 45.5, 118.6, 3.2, 0.0, 88.9]

# แบบเดิม (ลูป)
heavy = []
for v in rain:
    if v > 35:
        heavy.append(v)
print(heavy)

# แบบ comprehension
heavy2 = [v for v in rain if v > 35]
print(heavy2)

# แปลงหน่วย mm -> inch ทุกค่า
inches = [round(v / 25.4, 2) for v in rain]
print(inches)

# ใช้ร่วมกับ enumerate เพื่อเอา "วันที่" ที่ฝนหนัก
heavy_days = [i + 1 for i, v in enumerate(rain) if v > 35]
print("วันที่ฝนหนัก:", heavy_days)
````

## 4. Nested list — ตารางข้อมูล

list ซ้อน list ใช้แทนตาราง 2 มิติได้ `table[แถว][คอลัมน์]`
โครงสร้างนี้คือต้นแบบของ **raster grid** ในงาน GIS และของ DataFrame ที่จะเรียนวันที่ 5

````python
# [station_id, lat, lon, annual_rain_mm]
table = [
    ["CHB01", 13.367, 100.983, 1080.5],
    ["SRC01", 13.163, 100.803,  946.6],
    ["NYI01", 13.150, 101.350, 1435.8],
    ["STH01", 12.680, 100.983, 1199.4],
]

print(table[0])
print(table[0][0], table[0][3])

print(f"{'ID':<8}{'LAT':>9}{'LON':>10}{'RAIN':>10}")
for row in table:
    print(f"{row[0]:<8}{row[1]:>9.3f}{row[2]:>10.3f}{row[3]:>10.1f}")
````

````python
# ✍️ พิมพ์โค้ดด้านบนด้วยตัวเองที่นี่
````

````python
# หาสถานีที่ฝนมากที่สุด
wettest = table[0]
for row in table:
    if row[3] > wettest[3]:
        wettest = row
print("ฝนมากที่สุด:", wettest[0], wettest[3], "mm")

# เรียงตามปริมาณฝนจากมากไปน้อย (key บอกว่าใช้คอลัมน์ไหนเรียง)
for row in sorted(table, key=lambda r: r[3], reverse=True):
    print(f"{row[0]} {row[3]:.1f}")
````

````python
# ✍️ พิมพ์โค้ดด้านบนด้วยตัวเองที่นี่
````

## 5. Tuple — list ที่แก้ไม่ได้

เขียนในวงเล็บกลม `( )` **แก้ค่าไม่ได้หลังสร้าง (immutable)**

ใช้กับข้อมูลที่ "ไม่ควรเปลี่ยน" เช่น **พิกัด** — ถ้าเผลอเขียนทับ lat ของจุดใดจุดหนึ่ง
ข้อมูลทั้งชุดจะผิดโดยไม่มีใครรู้ตัว tuple ช่วยกันพลาดตรงนี้

````python
point = (13.367, 100.983)
print(point, point[0], point[1])

lat, lon = point          # tuple unpacking
print(f"lat={lat}, lon={lon}")

# tuple แก้ไม่ได้ -> เซลล์นี้จะ error (ตั้งใจให้เห็น)
point[0] = 99.0
````

````python
# ✍️ ลองสร้าง tuple ของตัวเอง แล้ว unpack ออกมา
````
