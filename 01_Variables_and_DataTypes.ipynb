## 1. ตัวแปร (variable)

ตัวแปรคือ "กล่องที่มีชื่อ" ใช้เก็บค่าไว้ใช้ต่อ เครื่องหมาย `=` คือ **การกำหนดค่า** (assignment)

````python
station_name = "ชลบุรี"
station_id = "CHB01"
latitude = 13.367
longitude = 100.983
elevation_m = 3
is_coastal = True

print(station_name)
print(latitude, longitude)
````

## 3. การแปลงชนิดข้อมูล (type casting)

ข้อมูลที่อ่านจากไฟล์หรือจากผู้ใช้ **เป็น `str` เสมอ** ต้องแปลงก่อนคำนวณ

````python
rain_text = "45.5"          # ข้อความ ไม่ใช่ตัวเลข
print(rain_text + rain_text)    # str + str = ต่อข้อความ ไม่ใช่บวก!

rain_value = float(rain_text)   # แปลงเป็น float
print(rain_value + rain_value)  # ตอนนี้บวกได้จริง

print(int(45.9))    # int() ตัดทศนิยมทิ้ง ไม่ใช่ปัดเศษ -> 45
print(round(45.9))  # ถ้าต้องการปัดเศษใช้ round() -> 46
print(str(45.9) + " mm")
````

````python
# ✍️ พิมพ์โค้ดด้านบนด้วยตัวเองที่นี่
````

**ระวัง** ถ้าแปลงข้อความที่ไม่ใช่ตัวเลข จะเกิด `ValueError`
ปัญหานี้เจอบ่อยมากตอนอ่านไฟล์ข้อมูลจริงที่มีช่องว่างหรือคำว่า `NA` ปนอยู่
เดี๋ยววันที่ 5 เราจะเรียนวิธีจัดการ (`try` / `except`)

````python
print(float("ไม่มีข้อมูล"))
````

## 4. f-string — แสดงผลให้อ่านง่าย

เติม `f` หน้าเครื่องหมายคำพูด แล้วใส่ชื่อตัวแปรในวงเล็บปีกกา `{ }`

````python
rain_mm = 45.4789

print(f"สถานี {station_name} ({station_id})")
print(f"พิกัด {latitude}, {longitude}")
print(f"ฝนวันนี้ {rain_mm} mm")

# จัดรูปแบบตัวเลข: .2f = ทศนิยม 2 ตำแหน่ง
print(f"ฝนวันนี้ {rain_mm:.2f} mm")
print(f"ฝนวันนี้ {rain_mm:.1f} mm")
print(f"พิกัด {latitude:.3f}N {longitude:.3f}E")

# จัดความกว้างคอลัมน์: >8 = ชิดขวากว้าง 8 ช่อง
print(f"|{station_id:>8}|{rain_mm:>8.1f}|")
````

````python
# ✍️ พิมพ์โค้ดด้านบนด้วยตัวเองที่นี่
````

## 5. `input()` — รับค่าจากผู้ใช้

`input()` จะหยุดรอให้พิมพ์ แล้วคืนค่ามาเป็น **`str` เสมอ** ถ้าจะคำนวณต้อง `float()` ก่อน

````python
name = input("ชื่อของคุณ: ")
rain_str = input("ปริมาณฝนวันนี้ (mm): ")

rain = float(rain_str)
print(f"สวัสดีคุณ {name} ฝนวันนี้ {rain:.1f} mm คิดเป็น {rain/25.4:.2f} นิ้ว")
````

````python
# ✍️ พิมพ์โค้ดด้านบนด้วยตัวเองที่นี่
````
