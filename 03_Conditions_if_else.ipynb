## 2. `if` / `else` และ `elif`

- `else` = กรณีอื่นทั้งหมด
- `elif` = ตรวจเงื่อนไขถัดไป (ย่อมาจาก else if) ใส่กี่ตัวก็ได้

**Python จะตรวจจากบนลงล่าง แล้วหยุดที่เงื่อนไขแรกที่เป็นจริง** — ลำดับจึงสำคัญมาก

````python
rain_mm = 45.5

if rain_mm >= 0.1:
    print("มีฝน")
else:
    print("ไม่มีฝน")
````

````python
# ✍️ พิมพ์โค้ดด้านบนด้วยตัวเองที่นี่
````

### เกณฑ์ปริมาณฝนสะสม 24 ชั่วโมง (ใช้กันทั่วไปในไทย)

| ปริมาณฝน (mm) | ระดับ |
|---|---|
| < 0.1 | ไม่มีฝน |
| 0.1 – 10.0 | ฝนเล็กน้อย |
| 10.1 – 35.0 | ฝนปานกลาง |
| 35.1 – 90.0 | ฝนหนัก |
| > 90.0 | ฝนหนักมาก |

````python
rain_mm = 118.6

if rain_mm < 0.1:
    level = "ไม่มีฝน"
elif rain_mm <= 10.0:
    level = "ฝนเล็กน้อย"
elif rain_mm <= 35.0:
    level = "ฝนปานกลาง"
elif rain_mm <= 90.0:
    level = "ฝนหนัก"
else:
    level = "ฝนหนักมาก"

print(f"ฝน {rain_mm} mm -> {level}")
````

````python
# ✍️ พิมพ์โค้ดด้านบนด้วยตัวเองที่นี่
````

### ⚠️ กับดักที่พบบ่อย: เรียงเงื่อนไขผิด

ลองรันเซลล์ข้างล่าง แล้วคิดว่าทำไมฝน 118.6 mm ถึงถูกจัดเป็น "ฝนเล็กน้อย"

````python
rain_mm = 118.6

if rain_mm > 0.1:
    level = "ฝนเล็กน้อย"     # เงื่อนไขนี้จริงตั้งแต่แรก -> หยุดทันที
elif rain_mm > 35.0:
    level = "ฝนหนัก"
else:
    level = "ไม่มีฝน"

print(level, "<- ผิด! เพราะเรียงจากเงื่อนไขกว้างไปแคบ")
````

**บทเรียน:** เรียงจากเงื่อนไข **แคบที่สุด/รุนแรงที่สุดก่อน** หรือไล่ค่าจากน้อยไปมากอย่างเป็นระบบ
แล้วทดสอบด้วยค่าที่อยู่ตรง **ขอบเขต** เสมอ (0, 0.1, 10.0, 10.1, 35.0, 90.0, 90.1)

## 3. Nested if และ logical operator

ถ้าต้องตรวจหลายเงื่อนไขพร้อมกัน เลือกได้ 2 แบบ — ใช้ `and`/`or` มักอ่านง่ายกว่าซ้อน `if`

````python
rain_mm = 92.0
elevation_m = 45
is_slope = True

# แบบซ้อน
if rain_mm > 90:
    if is_slope:
        print("เสี่ยงดินถล่มสูง")
    else:
        print("เสี่ยงน้ำท่วมขัง")

# แบบใช้ and (อ่านง่ายกว่า)
if rain_mm > 90 and is_slope:
    print("เสี่ยงดินถล่มสูง")
````

````python
# ✍️ พิมพ์โค้ดด้านบนด้วยตัวเองที่นี่
````

## 4. เขียนสั้นด้วย conditional expression

`ค่าเมื่อจริง if เงื่อนไข else ค่าเมื่อเท็จ` — ใช้เมื่อเงื่อนไขสั้น ๆ เท่านั้น
ถ้าซับซ้อนให้เขียน `if` เต็มรูปแบบ อ่านง่ายกว่า

````python
rain_mm = 2.5
status = "ฝนตก" if rain_mm >= 0.1 else "ไม่มีฝน"
print(status)
````

````python
# ✍️ พิมพ์โค้ดด้านบนด้วยตัวเองที่นี่
````
