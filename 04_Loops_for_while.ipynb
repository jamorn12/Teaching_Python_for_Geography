## 1. `for` กับ `range()`

`range(stop)` / `range(start, stop)` / `range(start, stop, step)`
สร้างลำดับตัวเลข **ไม่รวมค่า stop** (เหมือน slicing)

````python
for i in range(5):
    print("รอบที่", i)

print("---")
for month in range(1, 13):
    print(month, end=" ")

print()
for i in range(0, 101, 25):
    print(i, end=" ")
````

## 3. Accumulator pattern — รวม / นับ / หาค่าสูงสุด

รูปแบบนี้คือหัวใจของการวิเคราะห์ข้อมูลทั้งหมด:
**เตรียมตัวแปรเก็บผลไว้ก่อนลูป → อัปเดตในลูป → ใช้ผลหลังลูป**

````python
rain = [0.0, 12.4, 0.0, 45.5, 118.6, 3.2, 0.0, 88.9]

total = 0          # 1) เตรียมไว้ก่อน
rain_days = 0
max_rain = 0
max_index = -1

for i in range(len(rain)):   # 2) อัปเดตในลูป
    value = rain[i]
    total = total + value
    if value >= 0.1:
        rain_days = rain_days + 1
    if value > max_rain:
        max_rain = value
        max_index = i

# 3) ใช้ผลหลังลูป
print(f"ฝนรวม        = {total:.1f} mm")
print(f"จำนวนวันฝนตก = {rain_days} วัน")
print(f"ฝนสูงสุด     = {max_rain:.1f} mm (วันที่ {max_index + 1})")
print(f"เฉลี่ยต่อวัน  = {total / len(rain):.2f} mm")
````

````python
# ✍️ พิมพ์โค้ดด้านบนด้วยตัวเองที่นี่
````

> Python มีคำสั่งลัด `sum()`, `max()`, `min()`, `len()` ให้ใช้อยู่แล้ว
> แต่ต้องเขียนลูปเองให้เป็นก่อน เพราะงานจริงส่วนใหญ่ซับซ้อนกว่าที่คำสั่งลัดทำได้

````python
print(sum(rain), max(rain), min(rain), len(rain))
````

## 4. `enumerate()` และ `zip()`

- `enumerate()` ให้ทั้ง **ลำดับ** และ **ค่า** พร้อมกัน
- `zip()` วนสองชุดข้อมูลไปพร้อมกัน

````python
for i, value in enumerate(rain):
    if value > 35:
        print(f"วันที่ {i + 1}: {value} mm -> ฝนหนัก")

print("---")
stations = ["CHB01", "SRC01", "NYI01", "STH01"]
annual = [1080.5, 946.6, 1435.8, 1199.4]

for name, total in zip(stations, annual):
    print(f"{name:>6} : {total:8.1f} mm")
````

````python
# ✍️ พิมพ์โค้ดด้านบนด้วยตัวเองที่นี่
````

## 5. `while`, `break`, `continue`

- `while` ใช้เมื่อ **ไม่รู้จำนวนรอบ** ล่วงหน้า ต้องมีเงื่อนไขที่ทำให้จบ ไม่งั้นวนไม่รู้จบ
- `break` = ออกจากลูปทันที
- `continue` = ข้ามไปรอบถัดไป

````python
# ฝนสะสมถึงเท่าไรจึงจะครบ 100 mm
rain = [0.0, 12.4, 0.0, 45.5, 118.6, 3.2, 0.0, 88.9]

total = 0
day = 0
while total < 100 and day < len(rain):
    total += rain[day]
    day += 1

print(f"ฝนสะสมครบ 100 mm ในวันที่ {day} (สะสม {total:.1f} mm)")
````

````python
# ✍️ พิมพ์โค้ดด้านบนด้วยตัวเองที่นี่
````

````python
for i, value in enumerate(rain):
    if value == 0.0:
        continue            # ข้ามวันที่ไม่มีฝน
    if value > 100:
        print(f"พบฝนหนักมากวันที่ {i+1} -> หยุดตรวจ")
        break
    print(f"วันที่ {i+1}: {value} mm")
````

````python
# ✍️ พิมพ์โค้ดด้านบนด้วยตัวเองที่นี่
````

## 6. Nested loop — ลูปซ้อนลูป

ใช้กับข้อมูลสองมิติ เช่น หลายสถานี × หลายเดือน (ซึ่งคือหน้าตาของข้อมูล grid ในงาน GIS)

````python
stations = ["CHB01", "NYI01"]
monthly = {
    "CHB01": [0.0, 0.0, 10.0, 38.2, 120.5, 150.1],
    "NYI01": [1.2, 0.0, 18.6, 52.0, 168.3, 175.4],
}

for s in stations:
    print(f"สถานี {s}")
    for m in range(6):
        print(f"   เดือน {m+1:>2} : {monthly[s][m]:6.1f} mm")
````

````python
# ✍️ พิมพ์โค้ดด้านบนด้วยตัวเองที่นี่
````
