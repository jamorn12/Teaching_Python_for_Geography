## 1. Dictionary — ข้อมูลที่เรียกด้วย "ชื่อ" ไม่ใช่ "ตำแหน่ง"

เขียนด้วย `{ key: value }` เข้าถึงด้วย `d[key]`

เทียบกับ list: list ต้องจำว่าปริมาณฝนอยู่คอลัมน์ที่ 3 แต่ dict เรียก `station["rain_mm"]` ได้เลย
โค้ดอ่านง่ายกว่ามาก และนี่คือหน้าตาของข้อมูล **GeoJSON** ที่ใช้กันทั่วไปในงาน GIS

````python
station = {
    "id": "CHB01",
    "name": "ชลบุรี",
    "lat": 13.367,
    "lon": 100.983,
    "elevation_m": 3,
    "annual_rain_mm": 1080.5,
}

print(station["name"])
print(station["lat"], station["lon"])
print(len(station))
````

````python
# ✍️ พิมพ์โค้ดด้านบนด้วยตัวเองที่นี่
````

### เพิ่ม แก้ ลบ และ `get()`

````python
station["province"] = "ชลบุรี"        # เพิ่ม key ใหม่
station["annual_rain_mm"] = 1100.0    # แก้ค่าเดิม
print(station)

del station["elevation_m"]            # ลบ
print("elevation_m" in station)

# print(station["temp"])              # <- key ที่ไม่มี จะ KeyError
print(station.get("temp"))            # get() คืน None แทนที่จะพัง
print(station.get("temp", -999))      # กำหนดค่า default ได้
````

````python
# ✍️ พิมพ์โค้ดด้านบนด้วยตัวเองที่นี่
````

## 2. วนลูปบน dictionary

````python
for key in station:
    print(key, "=", station[key])

print("---")
for key, value in station.items():
    print(f"{key:>16} : {value}")

print("---")
print(list(station.keys()))
print(list(station.values()))
````

````python
# ✍️ พิมพ์โค้ดด้านบนด้วยตัวเองที่นี่
````

## 3. Nested structure — dict ซ้อน dict / dict ของ list

โครงสร้างนี้ใช้แทนฐานข้อมูลหลายสถานีได้เลย

````python
db = {
    "CHB01": {"name": "ชลบุรี",   "lat": 13.367, "lon": 100.983, "rain": 1080.5},
    "SRC01": {"name": "เกาะสีชัง", "lat": 13.163, "lon": 100.803, "rain":  946.6},
    "NYI01": {"name": "หนองใหญ่",  "lat": 13.150, "lon": 101.350, "rain": 1435.8},
}

print(db["NYI01"]["name"])
print(db["NYI01"]["rain"])

for sid, info in db.items():
    print(f"{sid} {info['name']:<12} ฝนรวม {info['rain']:8.1f} mm")

# หาสถานีที่ฝนมากที่สุด
wettest = max(db, key=lambda k: db[k]["rain"])
print("ฝนมากที่สุด:", wettest, db[wettest]["name"])
````

````python
# ✍️ พิมพ์โค้ดด้านบนด้วยตัวเองที่นี่
````

## 4. dict เป็นตัวนับ (counter pattern)

ใช้บ่อยมาก เช่น นับจำนวนวันฝนตกแยกตามระดับ หรือนับเหตุการณ์แยกตามจังหวัด

````python
rain = [0.0, 12.4, 0.0, 45.5, 118.6, 3.2, 0.0, 88.9, 22.1, 0.5, 36.7, 95.2]

def level_of(mm):
    if mm < 0.1:
        return "ไม่มีฝน"
    elif mm <= 10:
        return "ฝนเล็กน้อย"
    elif mm <= 35:
        return "ฝนปานกลาง"
    elif mm <= 90:
        return "ฝนหนัก"
    return "ฝนหนักมาก"

counter = {}
for v in rain:
    lv = level_of(v)
    counter[lv] = counter.get(lv, 0) + 1   # ถ้ายังไม่มี key ให้เริ่มที่ 0

for lv, n in counter.items():
    print(f"{lv:<12} {n:>3} วัน  {'#' * n}")
````

````python
# ✍️ พิมพ์โค้ดด้านบนด้วยตัวเองที่นี่
````

## 5. Set — ชุดของค่าที่ไม่ซ้ำ

เขียนด้วย `{ }` (ไม่มี key) หรือ `set()` คุณสมบัติสำคัญ: **ไม่มีค่าซ้ำ และไม่มีลำดับ**

| ตัวดำเนินการ | ความหมาย |
|---|---|
| `a | b` | union — รวมกัน |
| `a & b` | intersection — ที่มีร่วมกัน |
| `a - b` | difference — มีใน a แต่ไม่มีใน b |

````python
provinces = ["ชลบุรี", "ระยอง", "ชลบุรี", "จันทบุรี", "ระยอง", "ชลบุรี"]

unique = set(provinces)
print(unique, len(unique))

have_data_2024 = {"CHB01", "SRC01", "NYI01"}
have_data_2025 = {"CHB01", "NYI01", "STH01", "PTY01"}

print("มีทั้งสองปี  :", have_data_2024 & have_data_2025)
print("มีอย่างน้อย 1 ปี:", have_data_2024 | have_data_2025)
print("หายไปในปี 2025:", have_data_2024 - have_data_2025)
print("เพิ่มมาใหม่   :", have_data_2025 - have_data_2024)
````

````python
# ✍️ พิมพ์โค้ดด้านบนด้วยตัวเองที่นี่
````
