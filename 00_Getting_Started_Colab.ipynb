## 1. Colab คืออะไร

Google Colab คือ Jupyter Notebook ที่รันอยู่บนเครื่องของ Google
เราไม่ต้องติดตั้ง Python เอง แค่มีบัญชี Google + เบราว์เซอร์ ก็เขียนโค้ดได้ทันที

โครงสร้างของ notebook มีแค่ 2 อย่าง:

| ชนิดของ cell | ใช้ทำอะไร | เพิ่มด้วย |
|---|---|---|
| **Code cell** | เขียนโค้ด Python แล้วรัน | `Ctrl` + `M` แล้ว `B` |
| **Text cell** | เขียนคำอธิบาย (Markdown) | `Ctrl` + `M` แล้ว `M` |

**คีย์ลัดที่ต้องจำวันนี้**

| คีย์ | ความหมาย |
|---|---|
| `Shift` + `Enter` | รัน cell นี้ แล้วไป cell ถัดไป |
| `Ctrl` + `Enter` | รัน cell นี้ แล้วอยู่ที่เดิม |
| `Ctrl` + `M` `B` | แทรก cell ใหม่ด้านล่าง |
| `Ctrl` + `M` `D` | ลบ cell |
| `Ctrl` + `/` | comment / uncomment บรรทัดที่เลือก |

> ก่อนเริ่ม ให้กด **File → Save a copy in Drive** เพื่อให้ได้ notebook ของตัวเอง
> ถ้าไม่ทำ งานที่พิมพ์จะหายเมื่อปิดหน้าต่าง

## 4. runtime และ shell command

- **Runtime** = เครื่องเสมือนที่กำลังรันโค้ดเราอยู่ ตัวแปรทั้งหมดเก็บอยู่ในนั้น
- ถ้าโค้ดค้าง หรือตัวแปรมั่ว → **Runtime → Restart session** แล้วรันใหม่ตั้งแต่ cell แรก
